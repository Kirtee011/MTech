#### Question 1. Define a queueing system.

A **queueing sustem** is a system or method used to manage and process waiting lines of customers (people,tasks, data, machines, jobs, etc). Key components include:
1. _Arrival process_: describes how customers arrive 
2. _Service mechanishm_: capacity, configuration and rate of the servers at which customers are served
3. _Queue Discipline_: the rule governing who arrives next (eg. FIFO, LIFO, priority based)
4. _Number of servers_: the number of servers available to provide the service

#### Question 2. What does M/M/1 represent in queueing theory?

**Kendall's Notation** is a shorthand code to represent key components of a queuing system. It is a three code system given by: \
**A/S/c** \
where, \
A = arrival distribution \
S = service distribution \
c = number of servers 

**M/M/1** in queueing system represents a queue with Markovian  (Poisson arrivals), Markovidan (Exponential service times) and a One server.

**Arrival Process (λ): M** : Rate of requests/tasks (e.g., data streams): follows Poisson distribution \
**Service Process (μ): M** : Rate of task completion (e.g., model inference speed): follows exponential distribution \
**Server**: Single Server (1)

M/M/1 queue model assumes that there is infinite population, infinite queue capacity and customers are served on FCFS (First Come First Serve Basis).

#### Question 3. Differentiate between:
* Arrival rate and service rate
* Queue length and system length
* Waiting time in queue and waiting time in system


 **Difference between arrival rate and service rate** 
 
 **Arrival rate(λ)** \
 -> The average number of customers arriving per unit time. Denoted by λ. \
 -> It measures customer entering the system. \
 -> It denotes demand. 

 **Service Rate (μ)** \
 -> The average number of customers served per unit time. Denoted by μ. \
 -> It measures customer leaving the system. \
 -> It denotes capacity. 

**Difference between queue length and system length** 

**Queue length (Lq)** \
-> Average number of customers waiting in the queue for service. Denoted by Lq. \
-> Does not include the customer being served. 

$L_q = \frac{\lambda^2}{\mu(\mu - \lambda)}$

**System length(L)** \
-> The number of customer in the entire system. \
-> Includes the customer waiting as well as customers being served.

$L = \frac{\lambda}{\mu - \lambda} = \frac{\rho}{1 - \rho}$

**Difference between waiting time in queue and waiting time in system** 

**Waiting time in queue(Wq)** \
-> Time spent waiting before service begins. \
-> Does not include service time.

$W_q = \frac{L_q}{\lambda}$, $W_q = \frac{\lambda}{\mu(\mu - \lambda)}$ 

**Waiting time in system(W)** \
-> Total time spent in the system. \
-> Includes Waiting time + Service time

$W = \frac{L}{\lambda}$,   $W = \frac{1}{\mu - \lambda}$

#### Question 4: 

Customers arrive at a service center according to a Poisson process at an average rate of 8 customers per hour. The service time follows an exponential distribution with a mean service rate of 12 customers per hour.

Find:
1. Traffic intensity ($\rho$)
2. Probability that the system is empty ($P_0$)
3. Average number of customers in the system (L)
4. Average number of customers in the queue ($L_q$)
5. Average waiting time in the system ($W$)
6. Average waiting time in the queue ($W_q$)

Arrival rate ($\lambda$) = 8 

Service rate ($\mu$) = 12

Traffic intensity ($\rho$) = $\frac{\lambda}{\mu}$ = $\frac{4}{12}$ = 0.667

Probability that the system is empty ($P_0$) = 1 - $\rho$ =  1 - 2/3 = 0.33

Average number of customers in the system (L) = $\frac{\rho}{1 - \rho}$ = $\frac{0.667}{0.333}$ = 2.003

Average number of customers in the queue ($L_q$) = No of customers in the system(L) - utilization factor($\rho$) = 2.003 - 0.667 = 1.336 

Average waiting time in the system ($W$) = $\frac{L}{\lambda}$ = 2.003/8 = 0.25


Average waiting time in the queue ($W_q$) = $\frac{L_q}{\lambda}$  = 1.336/8 = 0.167

#### Question 5:

A bank counter receives customers at a rate of 15 customers per hour. The cashier can serve 20 customers per hour.

Calculate:

1. Probability of having exactly 3 customers in the system.
2. Probability of having more than 5 customers in the system.
3. Average queue length.
4. Average waiting time in queue.

Arraival rate ($\lambda$) = 15

Service rate ($\mu$) = 20

Utilization factor ($\rho$) = $\frac{\lambda}{\mu}$ = 15/20 = 0.75

Probability of having 5 customer in the system = $(1-\rho)\rho^n$ = $(1-0.75) * 0.75^3$ = 0.1054

Probability of having more than 5 customer in the system  = $\rho^{n+1}$ = $0.75^{5+1}$ = 0.1179 

Average queue length ($L_q$) = $\frac{\lambda^2}{\mu(\mu-\lambda)}$ = $\frac{15^2}{20(20-15)}$ = 2.25

Average waiting time in queue ($W_q$) = $\frac{L_q}{\lambda}$ = $\frac{2.25}{15}$ = 0.15 hour

#### Question 6:

In a computer laboratory, students arrive at a printer according to a Poisson process at the rate of 10 students per hour. The printer serves requests exponentially at the rate of 14 students per hour.

Determine:

1. Utilization factor.
2. Expected number of students waiting.
3. Expected waiting time before printing.
4. Probability that there are at least 4 students in the system.

Arraival rate ($\lambda$) = 10

Service rate ($\mu$) = 14

Utilization factor ($\rho$) = $\frac{\lambda}{\mu}$ = 10/14 = 0.714

Expected number of students waiting ($L_q$) = $\frac{\lambda^2}{\mu(\mu-\lambda)}$ = $\frac{10^2}{14(14-10)}$ = 1.785

Expected waiting time before printing ($W_q$) = $\frac{L_q}{\lambda}$ = $\frac{1.785}{10}$ = 0.178 hour

Probability of having at least 4 student = Probability of having more than 3 customer in the system \
Probability of having more than 4 customer in the system P(x>3) = $\rho^{n+1}$ = $0.714^{3+1}$ = 0.259

#### Question 7
An ATM machine serves customers with exponentially distributed service times at a mean rate of 18 customers per hour. Customers arrive randomly at a rate of 16 customers per hour.
Find:
1. The probability that an arriving customer has to wait
2. The average time spent in the system
3. The probability that the waiting time in queue exceeds 10 minutes
4. The probability that there are no customers in the system

Arrival rate ($\lambda$) = 16 customers/hour

Service rate ($\mu$) = 18 customers/hour

Utilization factor ($\rho$) = $\frac{\lambda}{\mu}$ = $\frac{16}{18}$ = 0.889

Probability that an arriving customer has to wait = Probability that the server is busy

$P(\text{wait}) = \rho = 0.889$

Average time spent in the system ($W$) = $\frac{1}{\mu-\lambda}$

$W = \frac{1}{18-16} = \frac{1}{2} = 0.5$ hour = 30 minutes

Probability that the waiting time in queue exceeds 10 minutes

$P(W_q > t) = \rho e^{-(\mu-\lambda)t}$

where $t = 10$ minutes = $\frac{10}{60} = \frac{1}{6}$ hour

$P(W_q > 10 \text{ min}) = 0.889 \times e^{-2(1/6)}$

$= 0.889 \times e^{-0.333}$
$= 0.889 \times 0.717$
$= 0.637$

Probability that there are no customers in the system ($P_0$)

$P_0 = 1-\rho$
$P_0 = 1-0.889$
$= 0.111$

#### Question 8
A single repair technician repairs machines in a factory. Machines fail according to a Poisson process at the rate of 5 failures per day. Repair times are exponentially distributed with a mean repair rate of 8 repairs per day.
Calculate:

1. Average number of failed machines in the system
2. Average number waiting for repair
3. Average downtime per machine
4. Probability that more than 4 machines are in the system

Arrival rate ($\lambda$) = 5 failures/day

Service rate ($\mu$) = 8 repairs/day

Utilization factor ($\rho$) = $\frac{\lambda}{\mu}$ = $\frac{5}{8}$ = 0.625

Average number of failed machines in the system ($L$) = $\frac{\lambda}{\mu-\lambda}$

$L = \frac{5}{8-5}$
$= \frac{5}{3}$
$= 1.667$ machines

Average number of machines waiting for repair ($L_q$) = $\frac{\lambda^2}{\mu(\mu-\lambda)}$

$L_q = \frac{5^2}{8(8-5)}$
$= \frac{25}{24}$
$= 1.042$ machines

Average downtime per machine ($W$) = $\frac{1}{\mu-\lambda}$

$W = \frac{1}{8-5}$
$= \frac{1}{3}$ day
$= 0.333$ day

Probability that more than 4 machines are in the system

$P(N > 4) = \rho^{4+1}$
$= (0.625)^5$
$= 0.0954$
